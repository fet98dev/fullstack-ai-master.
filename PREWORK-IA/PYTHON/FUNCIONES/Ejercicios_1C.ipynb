FACTORIAL
Implementa una función recursiva llamada factorial que calcule el factorial de
un número entero positivo. El factorial de un número n se define como el
producto de todos los números enteros positivos desde 1 hasta n.
(El factorial de 0 y de 1 es igual a 1)

```python
def factorial(numero):
    """Calcular el factorial de un numero entero n"""
    #Caso base
    if numero <= 1:
        return 1

    #Caso recursivo
    else:
        return numero * factorial(numero - 1)

print(factorial(3))
```

POTENCIA
Implementa una función recursiva llamada potencia que calcule el resultado
de elevar un número a una potencia dada. Puedes asumir que tanto el
número como la potencia son enteros no negativos.

```python
def potencia(base, exponente):
    #Caso base
    if exponente == 0:
        return 1
    #Caso recursivo: La base multiplicada por la potencia del exponente anterior
    else:
        return base * potencia(base,exponente - 1)

#Ejemplos de uso
print(potencia(2,3))#Imprimer 8 (2*2*2)
print(potencia(5,0))#Imprimer 1 (caso base)
print(potencia(3,2))#Imprimer 9 (3*3)
```

ÜBUNGEN VON GEMINI

```python
def cuenta_atras(numero):
    """Cuenta atras con recursividad"""
    #Caso base
    if numero == 0:
        return print("DESPEGUE!")
    #Caso recursivo
    print(numero)

    cuenta_atras(numero - 1)

cuenta_atras(10)
```
