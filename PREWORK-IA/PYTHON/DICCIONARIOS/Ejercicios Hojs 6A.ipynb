24. Crea un diccionario llamado "equipos" que contenga tres entradas. Cada entrada representa un equipo deportivo y tiene las claves "nombre" y "jugadores" con sus respectivos valores. Los valores de "jugadores" deben ser listas con los nombres de los jugadores. Recorre el diccionario e imprime el nombre del equipo y la lista de jugadores. 

```python
equipos = {
    "equipo1": {
    "nombre": "UDALMERIA", 
    "jugadores": ["crusat","barbero"],
},
    "equipo2": {
    "nombre": "FCB",
    "jugadores": ["messi","suarez"]
},
    "equipo3": {
    "nombre": "Madrid",
    "jugadores": ["bicho","ramos"],
},
}
#Recorremos el diccionario e imprimimos el nombre y la lista de jugadores.
for equipo,info in equipos.items():
    #Convertimos la lista de los jugadores en una cadena separa por comas para que se vea mas limpio.
    lista_jugadores = ", ".join(info["jugadores"])
    print(f"El equipo es: {info['nombre']} y los jugadores son: {lista_jugadores}.")
```

25. Agrega un nuevo equipo al diccionario "equipos" utilizando una nueva clave y valor. La lista de jugadores debe contener al menos tres nombres. Imprime el diccionario actualizado.

```python
equipos = {
    "equipo1": {
    "nombre": "UDALMERIA", 
    "jugadores": ["crusat","barbero"],
},
    "equipo2": {
    "nombre": "FCB",
    "jugadores": ["messi","suarez"]
},
    "equipo3": {
    "nombre": "Madrid",
    "jugadores": ["bicho","ramos"],
},
}

equipos["equipo4"] = {"nombre": "Atlhetic", "jugadores": ["william","aduriz","laporte"]}

for equipo, info in equipos.items():
    lista_jugadores = ", ".join(info['jugadores'])
    nombre_equipos = info['nombre']
    print(f"El nombre del equipo es: {nombre_equipos}. Los jugadores son: {lista_jugadores}")
```
