CLASE PERSONA
Define una clase Persona con atributos como nombre, edad y profesión.
Luego, crea varios objetos de esta clase y muestra su información

```python
class Persona:
    def __init__(self,nombre,edad,profesion):
        self.nombre = nombre
        self.edad = edad
        self.profesion = profesion

la_persona = Persona("Juan","45","Informatico")

print(la_persona.nombre)
print(la_persona.edad)
print(la_persona.profesion)
```

CALCULADORA BÁSICA
Crea una clase llamada "Calculadora" con métodos para sumar, restar,
multiplicar y dividir. Crea objetos de esta clase y realiza algunas operaciones
básicas.

```python
class Calculadora:
    def __init__(self,n1,n2):
        self.suma = n1 + n2
        self.resta = n1 - n2
        self.multiplicacion = n1 * n2
        self.division = n1 / n2

calcular = Calculadora(10,2)
print(calcular.suma)
print(calcular.resta)
print(calcular.multiplicacion)
print(calcular.division)
```

LIBRO
Crea una clase “Libro” con atributos como título, autor y año de publicación. Luego, crea varios objetos Libro y muestra su información.

```python
class Libro:
    def __init__(self,titulo,autor,year):
        self.titulo = titulo
        self.autor = autor
        self.year = year

libro_1 = Libro("El quijote","cervantes","1498")
libro_2 = Libro("El hombre en busca de sentido","Viktor Frank","1945")


print(libro_1.titulo,libro_1.autor,libro_1.year)
print(libro_2.titulo,libro_2.autor,libro_2.year)
```

RECTÁNGULO
Crea una clase “Rectangulo” con atributos de longitud y ancho. Implementa un método para calcular el área y el perímetro del rectángulo.

```python

class Rectangulo:
    def __init__(self,long,anch):
        self.longitud = long
        self.anch = anch

    def calcular_area(self):
        area = self.longitud * self.anch
        return area
    
    def calcular_perimetro(self):
        return 2 * (self.longitud + self.anch)

rectangulo1 = Rectangulo(5,3)
print("Area:", rectangulo1.calcular_area())
print("Perimetro:", rectangulo1.calcular_perimetro())

```
